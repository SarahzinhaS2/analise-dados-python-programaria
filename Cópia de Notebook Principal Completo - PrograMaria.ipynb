### Valores Discrepantes (outliers)

```python
lista_idades = [26,30,32,22,26,35,400,20,43,31,23] # Lista das idades com um valor discrepante (400)
```

```python
media = np.mean(lista_idades) # Atribuindo a média da lista de idades para a variável "media"
media # Retornando a variável "media"
```

```python
desvio = np.std(lista_idades) # Atribuindo o desvio padrão da lista de idades para a variável "desvio"
desvio # Retornando a variável "desvio"
  # É interessante observar aqui, que o desvio padrão da nossa lista é um número bem grande e estranho, principalmente para a nossa lista com poucos números
  # Isso deixa bem óbvio que existem valores discrepantes na nossa lista, mais conhecidos como "Outliers"
```

```python
media + 3*desvio # Um cálculo para desvendar os outliers. Esse é o limite superior dos nossos valores, tudo acima desse número é um outlier
```

```python
media - 3*desvio # Outra parte do cálculo para desvendar os outliers. Esse é o limite inferior dos nossos valores, tudo abaixo desse número é um outlier
```

```python
plt.boxplot(lista_idades) # Gerando um gráfico da nossa lista_idades usando a biblioteca plt
# A linha laranja no meio da caixa é a mediana
# As bordas da caixinha é nosso quartil Q1 e Q3
# A linha preta inferior é o nosso limite inferior, e a linha preta superior é nosso limite superior
  # Tudo além dessas linhas são nossos outliers
    # A bolinha no número "400" é um outlier, especificamente o valor 400 da nossa lista de idades
```

```python
plt.boxplot(dados["SALARIO"]) # Gerando um gráfico da coluna "Salário" usando a biblioteca plt
# Observamos que a caixa está mais ou menos nos "R$50.000", e que temos vários valores acima dos nossos limites, e todos são outliers
```

```python
Q1 = dados["SALARIO"].quantile(0.25) # Atribuindo o primeiro quartil usando a função "quartile" para a variável Q1. Retornamos apenas 25% do valor
Q1 # Retornando a variável Q1
```

```python
Q3 = dados["SALARIO"].quantile(0.75) # Atribuindo o terceiro quartil usando a função "quartile" para a variável Q3. Retornamos apenas 75% do valor
Q3 # Retornando a variável Q3
```

```python
IQR = Q3 - Q1 # Calculando o Interquartil para a variável IQR, que é o quartil entre os dois quartis que já calculamos
IQR # Retornando a variável IQR
```

```python
lim_superior = Q3 + (1.5*IQR) # Calculando o limite superior para a variável lim_superior
lim_superior # Retornando o limite superior
# Observasmos assim, que todos os valores acima de "22.359" são valores discrepantes/outliers
```

```python
lim_inferior = Q1 - (1.5*IQR) # Calculando o limite inferior para a variável lim_inferior
lim_inferior# Retornando o limite inferior
# Observasmos assim, que todos os valores abaixo de "0" ou "-5813" são valores discrepantes/outliers
  # Já que ninguém recebe salário negativo, podemos dizer que esse valor é "0" reais
```

# **- Correlação**

**Nesse módulo aprendi sobre Correlação, que é uma medida estatística que descreve a relação entre variáveis.**

Aprendi a usar algumas funções novas que me permitiram fazer uma correlação entre duas colunas. Esses cálculos variaram durante a aula, mas escrevi três formas diferentes de calcular correlação, todas são ótimas escolhas, e depende apenas da sua vontade escolher qual é a melhor para a sua análise específica dos dados. Tivemos dados que variam de `-1 à 1`, de `0 ao infinito` e de `0 a 1`.

Aprendi a usar o **Coeficiente de Crámer**, que é uma das formas de calcular essa correlação entre dois fatores, já que é uma forma mais complexa de calcular, deixei um pequeno resumo e várias anotações logo após o próximo código.

Também aprendi a gerar e ler **Tabelas Cruzadas** com e sem cabeçalho (Formato matriz/np.array). No formato matriz, fica tudo mais simples para início de machine learning, pois quanto menos texto e mais números puros, melhor.

Usei as funções `corr, pd.crosstab, np.array, chi2_contingency, np.sqrt, np.sum, np,min` e algumas funções que aprendi nos módulos anteriores.

```python
correlacao_continua = dados["IDADE"].corr(dados["SALARIO"])
# A função do pandas ".corr" faz o cáculo da correlação de dados contínuos
  # Quanto mais próximo de 1, a correlação é muito forte, então um afeta completamente o outro
  # Quanto mais próximo de -1, a correlação é muito fraca, então um não afeta em nada o outro
  # Quanto mais próximo de 0, a correlação é linear, logo não há diferença aparente entre as duas colunas
   # No nosso caso, estamos calculando a correlação entre as Idades e Salários, e atribuindo essa função para a variável correlacao_continua
   # Percebemos que a correlação é sim positiva, mas ela está mais próxima do 0 do que do 1, logo, a correlação não é muito forte entre esses dois parâmetros
correlacao_continua # Retornando a variável "correlacao_continua"
```

**No próximo código vamos usar o Coeficiente de Cramér, que é uma forma mais fácil de entender a correlação entre dois fatores.**
**Já que eu fiz muitas anotações, eu provavelmente terei dificuldades nesse tópico, então vou deixar uma explicação aqui sobre o que é o Coeficiente de Cramér:**

Para o cálculo de correlação entre variáveis categóricas, nós
podemos utilizar o Coeficiente de Cramér, que basicamente é a normalização de 0 a 1 da
correlação das variáveis, onde zero indica nenhuma associação entre as variáveis e 1 indica uma
associação completa. Para calcular o coeciente, vamos criar uma função que vai receber as
colunas que queremos a correlação.

```python
def cramer_coeficiente(coluna1,coluna2): # Atribuindo uma função chamada "cramer_coeficiente" que usa dois parâmetros (coluna1 e coluna2)
  tabela_cruzada = np.array(pd.crosstab(coluna1, coluna2))
    # Atribuindo a variável tabela_cruzada para uma função do pandas chamada ".crosstab", que faz uma Tabela Cruzada usando dois parâmetros
      # Um exemplo de crosstab está logo abaixo desse código!!!
    # Pegamos esse crosstab e usamos a função ".array" da biblioteca numpy para transformar essa tabela com cabeçalho em uma tabela no formato matriz
      # Um exemplo de array está abaixo da explicação de crosstab!!!
  chi2 = chi2_contingency(tabela_cruzada)[0]
    # O chi2_contingency (só com o chi2) mede o quanto sua tabela de dados é diferente do que seria esperado se não tivesse relação entre as coisas
      # No nosso caso pegamos apenas o primeiro valor do resultado usando o [0] (Interessante que o primeiro valor se chama propriamente chi2)
        # Vale gastar um tempo para ver os usos dessa função, pois eu dei uma pesquisada e é bem interessante
    # Depois de calcular o chi2, atribuimos o resultado para a variável "chi2"
  soma = np.sum(tabela_cruzada)
    # Fazendo a soma da tabela cruzada usando o numpy
  mini = np.min(tabela_cruzada.shape)-1
    # Fazendo o cálculo do valor mínimo da tabela.
      # Pegamos o valor mínimo do tamanho da tabela e subitraímos por 1
  cramer = np.sqrt(chi2/(soma*mini))
    # E finalmente podemos calcular o valor do coeficiente de cramer
      # O coeficiente de cramer é a raiz quadrada (np.sqrt) do chi2 dividido pela multiplicação da soma e do valor mínimo
    # Atribuimos esse cálculo para a variável "cramer"
  return cramer
    # Retornamos a função de cramer
```

**E para fácil acesso, deixarei essa mesma função no código abaixo, mas sem as anotações:**

```python
def cramer_coeficiente(coluna1, coluna2):
  tabela_cruzada = np.array(pd.crosstab(coluna1, coluna2))
  chi2 = chi2_contingency(tabela_cruzada)[0]
  soma = np.sum(tabela_cruzada)
  mini = np.min(tabela_cruzada.shape)-1
  cramer = np.sqrt(chi2/(soma*mini))
  return cramer
```
